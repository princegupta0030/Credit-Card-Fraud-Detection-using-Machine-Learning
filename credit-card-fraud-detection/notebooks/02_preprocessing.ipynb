# Data Preprocessing Pipeline

In this notebook, we demonstrate the usage of our custom preprocessing module located in `src/data_preprocessing.py`.

```python
import sys
import os

# Add the root directory to the python path so we can import from src
sys.path.append(os.path.abspath('..'))

from src.data_preprocessing import preprocess_pipeline
```

## 1. Run Preprocessing

We will now run the pipeline. This pipeline automatically:
1. Drops missing values and duplicate rows.
2. Drops the non-predictive `Time` column.
3. Separates the `Class` target variable from the features.
4. Splits the dataset using an 80/20 train/test split with stratification on the target class.
5. Scales the `Amount` feature. **Crucially, the scaler is fit ONLY on the training data to prevent data leakage.**

```python
raw_data_path = '../data/raw/creditcard.csv'
scaler_output_path = '../models/scaler.pkl'

X_train, X_test, y_train, y_test = preprocess_pipeline(
    filepath=raw_data_path, 
    scaler_output_path=scaler_output_path
)
```

## 2. Verify Output Shapes and Leakage Prevention

Let's verify that the shapes of the data match our expectations and that stratification maintained the fraud class ratio across train and test sets.

```python
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")

print(f"\nFraud ratio in training set: {y_train.mean():.4f}")
print(f"Fraud ratio in testing set: {y_test.mean():.4f}")
```

## 3. Verify Scaled Features

Let's check the mean and standard deviation of the `Amount` feature in the training and testing sets. Since the scaler was fit on the training set, its mean should be near 0 and std near 1. The test set will have slight variations, which is expected and confirms no data leakage.

```python
print("Training Set 'Amount' Stats:")
print(f"Mean: {X_train['Amount'].mean():.4f}, Std: {X_train['Amount'].std():.4f}")

print("\nTesting Set 'Amount' Stats:")
print(f"Mean: {X_test['Amount'].mean():.4f}, Std: {X_test['Amount'].std():.4f}")
```
